Filtering Rows And Columns

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Dell\Downloads\pandas_filtering_practice_dataset.csv")
df.head(5)
```

```python
#Q1:Select only the Customer and City columns.
df.filter(items=['Customer','City'])
```

```python
#Q2:Select Product, Price, and Rating columns.
df.filter(items=['Product','Price','Rating'])
```

```python
#Q3:Filter all orders where Price > 2000.
df[df['Price']>2000]
```

```python
#Q4:Filter all orders where Rating >= 4.
df[df['Rating']>=4]
```

```python
#Q5:Filter all orders where Quantity >= 3.
df[df['Quantity']>=3]
```

```python
#Q6:Show only orders where Order_Status == 'Delivered'.
df[df['Order_Status']=='Delivered']
```

```python
#Q7:Find orders where:Price > 2000 AND Rating >= 4
df[(df['Price']>2000) & (df['Rating']>=4)]
```

```python
#Q8:Find orders from Delhi OR Mumbai.
df[df['City'].isin (['Delhi','Mumbai'])]
```

```python
#Q9:Find orders where:Price is between 1000 and 3000 AND Rating >= 4
df[(df['Price'].between(1000,3000) & (df['Rating']>=4))]
```

```python
#Q10:Find Delivered orders where Total_Amount > 5000.
df[(df['Order_Status']=='Delivered') & (df['Total_Amount']>5000)]
```

```python
#Q11:Find orders from Delhi, Mumbai, or Hyderabad.
df[df['City'].isin(['Delhi','Mumbai','Hyderabad'])]
```

```python
#Q12:Find orders where:City is Delhi AND Price > 1500 AND Order_Status is Delivered
df[(df['City']=='Delhi') & (df['Price']>1500) & (df['Order_Status']=='Delivered')]
```

```python
#Q13:Show only Order_ID, Customer, Product, and Price for orders where Price > 2500.
df[df['Price']>2500][['Order_ID','Customer','Product','Price']]
```

```python
#Q14:Show Customer, City, and Total_Amount for customers from Delhi who have Rating >= 4.
df[(df['City']=='Delhi') & (df['Rating']>=4)][['Customer','City','Total_Amount']]
```

```python
#Q15:Show Order_ID, Product, Quantity, and Total_Amount for Delivered orders having Quantity >= 2.
df.loc[(df['Order_Status']=='Delivered') & (df['Quantity']>=2),['Order_ID','Product','Quantity','Total_Amount']]
```

```python
#Q16:Show Customer, City, Product, and Rating for orders where the city is Delhi or Mumbai and rating is at least 4.
df[(df['City'].isin(['Delhi','Mumbai'])) & (df['Rating']>=4)][[
    'Customer','City','Product','Rating']]
```

```python
'''Q17:Show Order_ID, Customer, Price, and Rating for orders where:
Price is between 1000–3000
Rating >= 4
Status = Delivered'''

df[df['Price'].between(1000,3000) & (df['Rating']>=4) &(df['Order_Status']=='Delivered')][[
    'Order_ID','Customer','Price','Rating']]
                                                                                          
```

```python
#Q18:Find all orders where the city is not Delhi and Rating >= 4.
df.loc[(df['City'] != 'Delhi') & (df['Rating'] >= 4),
    ['Customer', 'City', 'Rating']]
```

```python
'''Q19:Find orders where:
City is Delhi, Mumbai, or Hyderabad
Price is between 1000–3000
Rating >= 4
Status = Delivered'''

df.loc[ df['City'].isin(['Delhi','Mumbai','Hyderabad']) & df['Price'].between(1000, 3000) &
    (df['Rating'] >= 4) & (df['Order_Status'] == 'Delivered'),['Order_ID', 
    'Customer', 'Product', 'Total_Amount']]
     
```

```python
'''Q20:Final challenge: Return only:
Order_ID, Customer, City, Product, Quantity, Price, Rating, Total_Amount
for orders that satisfy all of these:Price >= 1500,Rating >= 4,
Quantity >= 2,City is Delhi or Mumbai,Order_Status is Delivered'''

df.loc[(df['City'].isin(['Delhi','Mumbai'])) & (df['Price']>=1500) &(df['Rating']>=4)&
    (df['Quantity']>=2) & (df['Order_Status']=='Delivered'),['Order_ID','Customer',
        'City','Product','Quantity','Price','Rating','Total_Amount']]
```
