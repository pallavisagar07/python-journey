Combining Dataframes

```python
import pandas as pd
```

```python
#Q1:Load both CSV files into separate DataFrames.
employee=pd.read_csv(r"C:\Users\Dell\Downloads\employees.csv")
employee
```

```python
performance=pd.read_csv(r"C:\Users\Dell\Downloads\performance.csv")
performance
```

```python
#Q2:Create a new DataFrame by concatenating two employee DataFrames vertically.
df_concat=pd.concat([employee,performance],axis=0)
df_concat
```

```python
#Q3:Concatenate them while resetting the index.
df_index= pd.concat([employee, performance], axis=0, ignore_index=True)
df_index
```

```python
#Q4:Concatenate two DataFrames horizontally and observe the result.
df_1 = pd.concat([employee, performance], axis=1)
df_1
```

```python
#Q5:Merge employees.csv and performance.csv using employee_id.
df_merge=pd.merge(employee,performance,on="employee_id")
df_merge
```

```python
#Q6:Perform a left merge using employee_id.
df_left=pd.merge(employee,performance,on="employee_id",how="left")
df_left
```

```python
#Q7:Perform a right merge using employee_id.
df_right=pd.merge(employee,performance,on="employee_id",how="right")
df_right
```

```python
#Q8:Perform an inner merge using employee_id
df_inner=pd.merge(employee,performance,on="employee_id",how="inner")
df_inner
```

```python
#Q9:Perform an outer merge using employee_id.
df=pd.merge(employee,performance,on="employee_id",how="outer")
df
```

```python
#Q10:After merging, calculate the total compensation = salary + bonus for every employee.
df["total_compensation"] = df["salary"] + df["bonus"]
df
```

```python
#Q11:Find the employee with the highest performance rating.
df.loc[df["rating"].idxmax()]
```

```python
#Q12:Find employees who have completed more than 3 projects.
df[df["projects"] > 3].groupby("name")["projects"].sum()
```

```python
#Q13:Find the average salary of employees whose rating is greater than 4.0.
df[df["rating"] > 4.0].groupby("name")["salary"].mean()
```

```python
#Q14:Find the department with the highest average salary.
df.groupby("department")["salary"].mean().sort_values(ascending=False).head(1)
```

```python
#Q15:Sort the merged DataFrame by rating from highest to lowest.
df.sort_values("rating",ascending=False)
```

```python
#Q16:Suppose a third CSV called attendance.csv.Load it and merge all three DataFrames using employee_id
attendance=pd.read_csv(r"C:\Users\Dell\Downloads\attendance.csv")
attendance
```

```python
df3=pd.merge(df_merge,attendance,on="employee_id")
df3
```

```python
#Q17:Calculate a new column:performance_score = rating × projects
df3["performance_score"]=df3["rating"]*df3["projects"]
df
```

```python
#Q18:Find the employee who has the highest performance score.
df3.loc[df3["performance_score"].idxmax()]
```

```python
#Q19:Find employees whose:rating > 4,projects > 3,days_present > 22
df3[(df3["rating"]>4)&(df3['projects']>3)&(df3['days_present']>22)]
```

```python
'''Q20:Create a final DataFrame containing only:
employee_id, name, department, salary, bonus, rating, projects, days_present, performance_score'''

df3 = df3[["employee_id","name","department","salary","bonus","rating","projects",
        "days_present","performance_score"]]
df3
```
