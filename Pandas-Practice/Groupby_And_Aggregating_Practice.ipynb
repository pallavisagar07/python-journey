Groupby And Aggregating

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Dell\Downloads\groupby_practice_ecommerce_dataset.csv")
df.head(5)
```

```python
#Q1:Find the total number of orders for each City.
df.groupby('City').size()
```

```python
#Q2:Find the total number of orders for each Category.
df.groupby('Category').size()
```

```python
#Q3:Find the number of orders for each Payment_Mode.
df.groupby('Payment_Mode').size()
```

```python
#Q4:Find the average Rating for each Category.
df.groupby('Category')['Rating'].mean()
```

```python
#Q5:Find the average Rating for each City.
df.groupby('City')['Rating'].mean()
```

```python
#Q6:Find the total Sales for each City.
df.groupby('City')['Sales'].sum()
```

```python
#Q7:Find the total Sales for each Category
df.groupby('Category')['Sales'].sum()
```

```python
#Q8:Find the average Sales for each Category.
df.groupby('Category')['Sales'].mean()
```

```python
#Q9:Find the maximum Sales for each City.
df.groupby('City')['Sales'].max()
```

```python
#Q10:Find the minimum Sales for each Category.
df.groupby('Category')['Sales'].min()
```

```python
#Q11:Find the total Sales for each City and Category.
df.groupby(['City','Category'])['Sales'].sum()
```

```python
#Q12:Find the average Rating for each City and Category.
df.groupby(['City','Category'])['Rating'].mean()
```

```python
#Q13:Find the number of orders for each City and Payment_Mode
df.groupby(['City','Payment_Mode']).size()
```

```python
#Q14:Find the total Sales for each Category and Payment_Mode.
df.groupby(['Category','Payment_Mode'])['Sales'].sum()
```

```python
#Q15:Find the number of orders for each Order_Status and Payment_Mode.
df.groupby(['Order_Status','Payment_Mode']).size()
```

```python
#Q16:Find the total sales and average sales for each Category.
df.groupby('Category')['Sales'].agg(['sum','mean'])
```

```python
#Q17:Find the minimum, maximum and average rating for each Category.
df.groupby('Category')['Rating'].agg(['min','max','mean'])
```

```python
#Q18:Find the total number of orders and total sales for each City.
df.groupby('City')['Sales'].agg(['size','sum'])
```

```python
#Q19:Find the number of orders for each Product, and identify which product has the highest number of orders.
order_count = df.groupby('Product').size()
order_count.sort_values(ascending=False)
```

```python
#Q20:Find the total sales for each City only for Delivered orders.
df[df['Order_Status']=='Delivered'].groupby('City')['Sales'].sum()
```

```python
#Q21:Find the top 5 cities by total sales.
df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(5)
```

```python
#Q22:Find the category with the highest average rating.
df.groupby('Category')['Rating'].mean().sort_values(ascending=False).head(1)
```

```python
#Q23:Find products having more than 10 orders
order_count = df.groupby('Product').size()
order_count[order_count > 10]
```

```python
#Q24:Find cities where total sales are greater than ₹50,000.
order_sales=df.groupby('City')['Sales'].sum()
order_sales[order_sales>50000]
```

```python
#Q25:Find the top 3 products by number of orders.
df.groupby('Product').size().sort_values(ascending=False).head(3)
```

```python
#Q26:Find the payment mode generating the highest sales.
df.groupby('Payment_Mode')['Sales'].sum().sort_values(ascending=False).head(1)
```

```python
#Q27:Find the category with the lowest average rating.
df.groupby('Category')['Rating'].mean().sort_values(ascending=True).head(1)
```

```python
#Q28:Find delivered orders and identify the city with the highest sales.
df[df['Order_Status']=='Delivered'].groupby('City')['Sales'].sum().sort_values(ascending=False).head(1)
```

```python
#Q29:Find the top 3 categories by total sales.
df.groupby('Category')['Sales'].sum().sort_values(ascending=False).head(3)
```

```python
#Q30:Find products whose average rating is greater than 4.
avg_rating=df.groupby('Product')['Rating'].mean()
avg_rating[avg_rating>4]
```
