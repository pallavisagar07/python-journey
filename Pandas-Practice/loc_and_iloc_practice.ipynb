```python
import pandas as pd
df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Neha'],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai'],
    'Sales': [5000, 7000, 3000, 8000, 6000],
    'Rating': [4.2, 4.5, 3.8, 4.7, 4.1]})
print(df)
```

```python
#Q1:Select Name and Sales of customers whose Sales are greater than 5000.
df.loc[df['Sales']>5000,['Name','Sales']]
```

```python
#Q2:Select customers who are from Mumbai AND have Sales greater than 6000.
df.loc[(df['City']=='Mumbai') & (df['Sales']>6000)]
```

```python
#Q3:Select only Name, City and Rating of customers whose Rating is greater than or equal to 4.2.
df.loc[df['Rating']>=4.2,['Name','City','Rating']]
```

```python
#Q4:Select customers who are from Delhi OR Pune.Return only Name and City.
df.loc[df['City'].isin(['Delhi','Pune']),['Name','City']]
```

```python
#Q5:Select rows at positions 1 through 3.
df.iloc[1:4]
```

```python
#Q6:Select rows at positions 1:4 and only columns at positions 0 and 2.
df.iloc[1:4, [0,2]]
```

```python
#Q7:Select Priya's Rating using loc.
df.loc[3,'Rating']
```

```python
#Q8:Select Neha's City using iloc.
df.iloc[4,1]
```

```python
#Q9:Select Name and Rating of customers who:have Sales ≥ 5000 AND Rating > 4.0
df.loc[(df['Sales']>=5000) &(df['Rating']>4.0),['Name','Rating']]
```

```python
#Q10:Select Name, Sales and Rating of customers who:are from Mumbai OR Delhi AND have Rating ≥ 4.1
df.loc[df['City'].isin(['Delhi', 'Mumbai']) & (df['Rating'] >= 4.1),['Name', 'Sales', 'Rating']]
```
