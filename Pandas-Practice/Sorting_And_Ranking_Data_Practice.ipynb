Sorting And Ranking Data

```python
import pandas as pd
df=pd.read_csv(r"C:\Users\Dell\Downloads\groupby_practice_ecommerce_dataset.csv")
df.head(5)
```

```python
#Q1:Sort the DataFrame by Sales in ascending order.
df.sort_values('Sales')
```

```python
#Q2:Sort the DataFrame by Sales in descending order.
df.sort_values('Sales',ascending=False)
```

```python
#Q3:Find the 5 orders with the highest Sales.
df.sort_values('Sales',ascending=False).head(5)
```

```python
#Q4:Find the 5 orders with the lowest Sales.
df.sort_values('Sales',ascending=True).head(5)
```

```python
#Q5:Sort the DataFrame by Rating from highest to lowest
df.sort_values('Rating',ascending=False)
```

```python
#Q6:Sort the DataFrame by Rating from lowest to highest.
df.sort_values('Rating',ascending=True)
```

```python
#Q7:Sort the DataFrame alphabetically by City.
df.sort_values('City')
```

```python
#Q8:Sort the DataFrame alphabetically by Product.
df.sort_values('Product')
```

```python
#Q9:Sort by City alphabetically and within each city sort Sales from highest to lowest.
df.sort_values(['City','Sales'],ascending=[True,False])
```

```python
#Q10:Sort by Category alphabetically and then by Sales in descending order.
df.sort_values(['Category','Sales'],ascending=[True,False])
```

```python
#Q11:Sort by Payment_Mode alphabetically and then by Sales from highest to lowest.
df.sort_values(['Payment_Mode','Sales'],ascending=[True,False])
```

```python
#Q12:Sort by Order_Status alphabetically and then by Rating from highest to lowest.
df.sort_values(['Order_Status','Rating'],ascending=[True,False])
```

```python
#Q13:Sort by City alphabetically and then by Rating from highest to lowest.
df.sort_values(['City','Rating'],ascending=[True,False])
```

```python
#Q14:Display only Delivered orders and sort them by Sales from highest to lowest.
df[df['Order_Status']=='Delivered'].sort_values('Sales',ascending=False)
```

```python
#Q15:Display only orders from Delhi and sort them by Sales descending.
df[df['City']=='Delhi'].sort_values('Sales',ascending=False)
```

```python
#Q16:Display orders where Sales > 1000 and sort them by Sales descending.
df[df['Sales']>1000].sort_values('Sales',ascending=False)
```

```python
#Q17:Display orders where Rating >= 4.5 and sort them by Rating descending.
df[df['Rating']>=4.5].sort_values('Rating',ascending=False)
```

```python
#Q18:Display Delivered orders with Rating >= 4.5, sorted by Sales from highest to lowest.
df[(df['Order_Status']=='Delivered')&(df['Rating']>=4.5)].sort_values('Sales',ascending=False)
```

```python
#Q19:Display orders from Mumbai and sort them by Rating from highest to lowest.
df[df['City']=='Mumbai'].sort_values('Rating',ascending=False)
```

```python
#Q20:Display orders where Payment_Mode == 'UPI' and sort them by Sales descending.
df[df['Payment_Mode']=='UPI'].sort_values('Sales',ascending=False)
```

```python
#Q21:Sort the DataFrame by its index in ascending order.
df.sort_index()
```

```python
#Q22:Sort the DataFrame by its index in descending order.
df.sort_index(ascending=False)
```

```python
#Q23:Set Order_ID as the index and sort the DataFrame by index
df.set_index('Order_ID').sort_index()
```

```python
#Q24:Set City as the index and sort the DataFrame by index.
df.set_index('City').sort_index()
```

```python
#Q25:Create a Sales_Rank column where the highest Sales gets rank 1.
df['Sales_Rank'] = df['Sales'].rank(ascending=False).astype(int)
df['Sales_Rank']

```

```python
#Q26:Create a Rating_Rank column where the highest Rating gets rank 1.
df['Rating_Rank']=df['Rating'].rank(ascending=False)
df['Rating_Rank']
df
```

```python
#Q27:Create a Sales_Rank column where the lowest Sales gets rank 1.
df['Sales_Rank'] = df['Sales'].rank(ascending=True)
df['Sales_Rank']
df
```

```python
#q28:Display the top 5 orders based on Sales rank.
df[df['Sales'].rank(ascending=False) <= 5]
```

```python
#Q29:Display all orders having a Sales_Rank <= 3.
df[df['Sales_Rank']<=3]
```

```python
#Q30:Create both Sales_Rank and Rating_Rank columns.
df['Sales_Rank'] = df['Sales'].rank(ascending=False)
df['Rating_Rank'] = df['Rating'].rank(ascending=False)
df
```

```python
#Q31:Find the highest-selling order in each City.
df.loc[df.groupby('City')['Sales'].idxmax()]
```

```python
#Q32:Find the highest-rated order in each City.
df.loc[df.groupby('City')['Rating'].idxmax()]
```
