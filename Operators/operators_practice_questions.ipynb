```python
#Q1:Write a program to input two numbers and print their sum,difference and product.

num1=int(input("Enter number :"))
num2=int(input("Enter number :"))
sum=num1+num2
diff=num1-num2
product=num1*num2
print("sum is:", sum)
print("difference is:", diff)
print("Product is:", product)
```
