Student Performance Analyzer — NumPy Project

```python
#.Import the libraries
import pandas as pd
import numpy as np
```

```python
#Task 1 — Load and Convert the Dataset
df=pd.read_csv(r"student_performance.csv")
df.head()
```

```python
#convert numerical data into numpy array
data = df[['Student_ID', 'Math', 'Science', 'English', 'Computer', 'Attendance']].to_numpy()
print(data)
```

```python
#Task 2 — Explore the Dataset
print("Shape:", data.shape)
print("Dimensions:", data.ndim)
print("Number of Students:", data.shape[0])
print("Total No. of Elements:", data.size)
```

```python
# Task 3 — Indexing and Slicing

print("First Student:", data[0])              # First student
print("Last Student:", data[-1])              # Last student
```

```python
print("First 5 Students:",data[:5])           # First 5 students
```

```python
print("Last 5 Students:",data[-5:])           # Last 5 students
```

```python
print("Math Scores:", data[:, 1])             # Math scores
print("Attendance:", data[:, 5])              #Attendance
```

```python
print("5th to 10th Students:",data[4:10])     # 5th to 10th students
```

```python
#Task 4 — Basic Statistical Analysis

scores = data[:, 1:]

print("Mean:", np.mean(scores, axis=0))
print("Median:", np.median(scores, axis=0))
print("Minimum:", np.min(scores, axis=0))
print("Maximum:", np.max(scores, axis=0))
print("Standard Deviation:", np.std(scores, axis=0))
print("Variance:", np.var(scores, axis=0))
```

```python
# Task 5 — Student Performance Classification

# Calculate average score of each student
subject_scores = data[:, 1:5]
average_score = np.mean(subject_scores, axis=1)

# Classify students
performance = np.where(average_score >= 80, "Excellent",
              np.where(average_score >= 60, "Good",
              np.where(average_score >= 40, "Average","Needs Improvement")))
print("Average Scores:", average_score)
print("Performance:", performance)
```

```python
# Task 6 — Attendance and Performance Analysis

# Average attendance
average_attendance = np.mean(data[:, 5])
print("Average Attendance:", average_attendance)

# Students with above-average attendance
above_average = data[:, 5] > average_attendance
print("Above Average Attendance:", above_average)

# Performance of students with above-average attendance
high_attendance_score = average_score[above_average]
print("Average Performance of High Attendance Students:",np.mean(high_attendance_score))

# Students with below-average attendance
below_average = data[:, 5] <= average_attendance

# Performance of students with below-average attendance
low_attendance_score = average_score[below_average]
print("Average Performance of Low Attendance Students:",np.mean(low_attendance_score))
```

```python
# Task 7 — Top and Bottom Performers

# Highest-performing student
top_index = np.argmax(average_score)
print("Top Student ID:", data[top_index, 0])
print("Top Student Score:", average_score[top_index])
```

```python
# Lowest-performing student
bottom_index = np.argmin(average_score)
print("Lowest Student ID:", data[bottom_index, 0])
print("Lowest Student Score:", average_score[bottom_index])
```

```python
# Top 3 performers
top_3_indices = np.argsort(average_score)[-3:][::-1]
print("Top 3 Students:")
print("Student IDs:", data[top_3_indices, 0])
print("Scores:", average_score[top_3_indices])
```

```python
# Bottom 3 performers
bottom_3_indices = np.argsort(average_score)[:3]
print("Bottom 3 Students:")
print("Student IDs:", data[bottom_3_indices, 0])
print("Scores:", average_score[bottom_3_indices])
```

```python
# Task 8 — Subject-wise Performance Analysis

scores = data[:, 1:5]
subjects = np.array(["Math", "Science", "English", "Computer"])

# Average of each subject
subject_means = np.mean(scores, axis=0)
print("Subject Means:", subject_means)
```

```python
# Highest average subject
highest_subject_index = np.argmax(subject_means)
print("Highest Average Subject:", subjects[highest_subject_index])
print("Average Score:", subject_means[highest_subject_index])
```

```python
# Lowest average subject
lowest_subject_index = np.argmin(subject_means)
print("Lowest Average Subject:", subjects[lowest_subject_index])
print("Average Score:", subject_means[lowest_subject_index])

# Overall average
overall_average = np.mean(scores)
print("Overall Average:", overall_average)
```

```python
# Students scoring 80 or above
count_above_80 = np.sum(scores >= 80, axis=0)
print("Students scoring 80 or above:", count_above_80)
```

```python
# Task 9 — Overall Performance Score

scores = data[:, 1:5]

# Calculate overall score for each student
overall_score = np.sum(scores, axis=1) / 4
print("Overall Score:", overall_score)
```

```python
# Class average
class_average = np.mean(overall_score)
print("Class Average:", class_average)

# Top-performing student
top_index = np.argmax(overall_score)
print("Top Student ID:", data[top_index, 0])
print("Top Overall Score:", overall_score[top_index])
```

```python
# Number of students above class average
above_class_average = overall_score > class_average
print("Students above class average:", np.sum(above_class_average))
```

```python
# Task 10 — Final Insights

print("Total Students:", data.shape[0])

print("Class Average:", np.mean(overall_score))

print("Highest Score:", np.max(overall_score))

print("Lowest Score:", np.min(overall_score))

excellent_students = np.sum(performance == "Excellent")
good_students = np.sum(performance == "Good")

print("Excellent Students:", excellent_students)
print("Good Students:", good_students)

highest_subject = subjects[np.argmax(subject_means)]
lowest_subject = subjects[np.argmin(subject_means)]

print("Highest Average Subject:", highest_subject)
print("Lowest Average Subject:", lowest_subject)

above_average_count = np.sum(overall_score > np.mean(overall_score))

print("Students Above Class Average:", above_average_count)
```

The analysis of 20 students shows an overall class average of 81.21. Most students performed well, with 13 students classified as Excellent and 7 as Good. Computer had the highest average performance among the subjects, while Math had the lowest. Additionally, 12 students scored above the overall class average, indicating generally strong academic performance in the dataset.
