```python
'''Q12:Write a program to calculate electricity bill:
Up to 100 units: ₹5/unit
101–200 units: ₹7/unit
Above 200 units: ₹10/unit'''
unit=int(input("Enter units"))
if unit<=100:
    Bill=5*unit
    print("Electricity bill:",Bill)
elif unit<=200:
    Bill=7*unit
    print("Electricity bill:",Bill)
elif unit>200:
    Bill=10*unit
    print("Electricity bill:",Bill)
```

```python
'''Q13:Write a program to check whether a character is:
Uppercase letter
Lowercase letter
Digit
Special character'''
str=input("Enter character:")
if str.islower():
    print("It is lowercase letter")
elif str.isupper():
    print("It is uppercase letter")
elif str.isdigit():
    print("It is digit")
else:
    print("It is special character")
```

```python
'''Q14:Write a program to determine the type of triangle based on three sides:
Equilateral
Isosceles
Scalene'''
len1=int(input("Enter length:"))
len2=int(input("Enter length:"))
len3=int(input("Enter length:"))
if len1==len2==len3:
    print("It is equilateral triangle")
elif len1==len2 or len1==len3 or len2==len3:
    print("It is isosceles triangle")
else:
    print("It is scalene triangle")
```

```python
'''Q16:Takes the user's account balance and withdrawal amount as input.
If the withdrawal amount is greater than the balance, print "Insufficient Balance".
Otherwise, deduct the amount and display the remaining balance.'''
balance=int(input("Enter account balance:"))
withdrawal=int(input("Enter withdrawal amount:"))
if withdrawal>balance:
     print("Insufficient balance")
else:
     rem_balance=balance-withdrawal
     print("Remaining balance is:", rem_balance)
```

```python
'''Q17:Write a program that calculates the delivery charge based on the order amount:

₹1000 or more → Free delivery
₹500–₹999 → ₹50 delivery charge
Below ₹500 → ₹100 delivery charge

Display the final payable amount.'''
order=int(input("Enter order amount:"))
if order>=1000:
    Delivery_charge = 0

elif order>=500:
    Delivery_charge = 50 

else :
       Delivery_charge = 100
Payable_amount=order+Delivery_charge
print("Delivery Charge:",Delivery_charge)
print("Total payable amount :",Payable_amount)
```
