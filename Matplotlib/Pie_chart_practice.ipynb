```python
'''1. Create a pie chart showing:
Python: 40
Java: 25
C++: 20
JavaScript: 15
Add a title and display percentages.'''
import matplotlib.pyplot as plt
languages=["Python","Java","C++","Javascript"]
y=[40,25,20,15]
plt.pie(y,labels=languages,autopct="%1.0f%%",radius=1)
plt.title("Favorite Programming Languages")
plt.show()

```

```python
'''Q4:A company sold:
Laptop: 120
Mobile: 250
Tablet: 80
Smartwatch: 150
Create a pie chart and display percentages.'''
import matplotlib.pyplot as plt
x=["Laptop","Mobile","Tablet","Smartwatch"]
y=[120,250,80,150]
plt.pie(y,labels=x,autopct="%1.0f%%")
plt.title("Company Sales by Product")
plt.show()
```

```python
'''Q5:Daily usage:
Instagram: 2.5 hours
YouTube: 2 hours
WhatsApp: 1 hour
LinkedIn: 0.5 hours
Other: 1 hour
Create a pie chart with a suitable title.'''
x=["Instagram","Youtube","Whatsapp","Linkedin","Other"]
y=[2.5,2,1,0.5,1]
color=["red","blue","yellow","maroon","magenta"]
plt.pie(y,labels=x,autopct="%1.0f%%",colors=color)
plt.title("Social Media Usage")
plt.show()
```

```python
'''Q6:6. Add an Exploded Slice
Using this data:
Python: 45
Java: 20
C++: 15
JavaScript: 20
Create a pie chart where Python is slightly separated (exploded) from the rest.'''
import matplotlib.pyplot as plt
languages=["Python","Java","C++","Javascript"]
y=[45,20,15,20]
color=["red","green","blue","yellow"]
plt.pie(y,labels=languages,autopct="%1.0f%%",explode=[0.1,0,0,0],radius=0.8,colors=color)
plt.title("Favorite Programming Languages")
plt.show()
```

```python
'''Q7:Create a pie chart for:
Renewable Energy: 35
Coal: 30
Natural Gas: 20
Nuclear: 10
Other: 5
Requirements:
Start the chart at a suitable angle.
Add a shadow.
Display percentages.'''
import matplotlib.pyplot as plt
x=["Coal","Natural Gas","Nuclear","Other"]
y=[30,20,10,5]
plt.pie(y,labels=x,autopct="%1.0f%%",startangle=45,shadow=True)
plt.title("Natural Resources")
plt.show()
```

```python
'''Q8:Create a pie chart showing:
Data Analysis: 30%
Machine Learning: 25%
Web Development: 20%
AI: 15%
Other: 10%

Requirements:
Use your own colors.
Display percentages.
Add a title.
Remove unnecessary borders if you know how.'''
x=["Data Analysis","Machine Learning","Web Development","AI","Other"]
y=[30,25,20,15,10]
color=["Green","skyblue","Yellow","Orange","red"]
plt.pie(y,labels=x,colors=color,radius=0.5,autopct="%1.0f%%",startangle=90)
plt.axis("equal")
plt.title("Course ")
plt.show()
```

```python
'''Q10:Suppose your LinkedIn post traffic came from:
Followers: 35%
Search: 20%
Home Feed: 30%
Profile Views: 10%
Other: 5%
Create a professional-looking pie chart.

Requirements:
Custom colors
Percentages
Exploded largest slice
Shadow
Suitable start angle
Title
Legend
Make the chart look suitable for posting on LinkedIn'''
import matplotlib.pyplot as plt
x=["Followes","Search","Home Feed","Profile Views","Other"]
y=[35,20,30,10,5]
color=["#32a862","#329ca8","#a83275","#a8a432","#a83e32"]
plt.pie(y,labels=x,autopct="%1.0f%%",startangle=90,colors=color,shadow=True,explode=[0.1,0,0,0,0])
plt.title("LinkedIn Analytics")
plt.legend(loc=2)
plt.show()
       
```
