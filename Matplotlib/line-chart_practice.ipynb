```python
import matplotlib.pyplot as plt
#Q1:Create a line chart showing the monthly sales of a store.
'''Requirements:
-Plot sales vs months.
-Add title.
-Label x-axis and y-axis.'''

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [1200, 1500, 1800, 1700, 2100, 2500]
plt.plot(months,sales)

#add title and axis labels
plt.title("Monthly sales of store")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#q2: Use the data below:
'''Requirements:
Use a red dashed line.
Add circle markers.
Increase line width.
Add grid.'''

days = [1, 2, 3, 4, 5, 6, 7]
temperature = [30, 32, 35, 33, 31, 29, 28]

#plot line chart
plt.plot(days,temperature, color="red", marker="o",linestyle="dashed",linewidth=2)

#add title and axis labels
plt.title("Daily temperature report")
plt.xlabel("Days")
plt.ylabel("Temperature")

plt.grid(True)              #add grid
plt.show()
         
```

```python
#Q3: Compare two products:
'''Requirements:

Plot both products on the same graph.
Add legend.
Use different colors and markers.'''

months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [40, 45, 50, 60, 70]
product_B = [35, 48, 55, 58, 65]

#plot both line chart
plt.plot(months,product_A, color="blue", marker="o",label="Product A")
plt.plot(months, product_B, color="red", marker="*",label="Product B")

#title and axis labels
plt.title("Comparison of Two Products")
plt.xlabel("Months")
plt.ylabel("Product")
plt.legend()              #add legend
plt.grid(True)                  #add grid
plt.show()
```

```python
'''q5:Requirements

Plot visitor trend.
Rotate x-axis labels.
Display grid.
Change figure size.'''

days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
visitors = [150,180,220,210,260,300,280]
plt.figure(figsize=(8,5))             #create figure

#plot line chart
plt.plot(days,visitors,color="red",marker="o")

#title and axis labels
plt.title("Daily Website Visitors")
plt.xlabel("Days")
plt.ylabel("Visitors")

plt.xticks(rotation=45)           #rotate x-axis labels
plt.grid(True)
plt.show()

```

```python
'''Q6:Requirements

Plot the graph.
Use green line.
Add square markers.
Set y-axis limit from 0 to 20.'''

import matplotlib.pyplot as plt

hours = list(range(0,11))
units = [2,3,4,6,8,10,11,13,15,16,18]
plt.figure(figsize=(7,4))
plt.plot(hours,units,color="green",marker="s",linewidth=2)
plt.title("Electricity Consumption")
plt.xlabel("Hours")
plt.ylabel("Units")
plt.ylim(0,20)              #set y-axis limit
plt.grid(True)
plt.show()
```

```python
#Q7: plot a line graph for stock price analysis
'''Requirements

Draw a blue line.
Highlight every point using markers.
Add legend.
Save the figure as stock_price.png.'''

days = [1,2,3,4,5,6,7]

price = [120,125,123,128,130,135,140]
plt.plot(days,price,color="blue", marker="o",label="Stock Price")
plt.title("Stock Price Analysis")
plt.xlabel("Days")
plt.ylabel(" Stock Price")
plt.legend()
plt.grid(True)
plt.savefig("stock_price.png")
plt.show()
```

```python
#Q8: plot a line graph for multiple cities temperature
'''Requirements

Plot all three cities.
Add legend.
Use different line styles.
Add title.'''
#data:
days = ["Mon","Tue","Wed","Thu","Fri"]
delhi = [36,37,38,39,40]
mumbai = [31,32,31,30,29]
bangalore = [27,28,29,28,27]
plt.plot(days,delhi, color="red",marker="o",label="Delhi temperature")
plt.plot(days,mumbai, color="blue",marker="s",linestyle="dotted",label="Mumbai temperature")
plt.plot(days,bangalore, color="green",marker="*",linestyle="dashed",label="Bangalore temperature")
plt.title("Weekly Cities Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Q9: plot a line chart for population growth
'''Requirements

Create a line graph.
Annotate the highest population point.
Add labels and title.
Display grid.'''

#Data:
years = [2018,2019,2020,2021,2022,2023]

population = [1.2,1.25,1.31,1.36,1.42,1.50]
plt.plot(years,population, color="blue",marker="o",linewidth=2)
plt.title("Population Growth(2018-2023)")
plt.xlabel("Years")
plt.ylabel("Population in millions")

#find highest population
max_population=max(population)
max_year=years[population.index(max_population)]

#annotate the highest point
plt.annotate("Highest Population",xy=(max_year,max_population),xytext=(2021,1.46),
             arrowprops=dict(arrowstyle="->", color="black"))
plt.grid(True)
plt.show()


```

```python
#Q10: plot a line chart for company revenue dashboard using given data
'''Requirements

Plot all three years.
Use different colors.
Add legend.
Add title.
Label axes.
Add markers.
Add grid.
Save the figure as company_revenue.png.
Display the graph.
Bonus Challenge (Interview Level)'''
#Data:
quarters = ["Q1","Q2","Q3","Q4"]
revenue_2022 = [25,30,28,35]
revenue_2023 = [30,35,38,42]
revenue_2024 = [38,42,46,50]
#plot graph
plt.plot(quarters,revenue_2022,color="red",marker="o", linewidth=2,label="2022 revenue")
plt.plot(quarters,revenue_2023,color="blue",marker="*",markersize=7, linewidth=2,label="2023 revenue")
plt.plot(quarters,revenue_2024,color="black",marker="s", linewidth=2,label="2024 revenue")
plt.title("Company Revenue Dashboard")
plt.xlabel("Quarters")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)
plt.savefig("company_revenue.png")
plt.show()
```
