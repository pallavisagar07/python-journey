```python
#q1:Create a bar chart using the given student names and marks 
students=["Aman","Riya","Karan","Neha","Priya"]
marks= [78,92,65,88,75]
 
#Add title: "Students Marks"
#Label x-axis and y-axis

import matplotlib.pyplot as plt
x=["Aman","Riya","Karan","Neha","Priya"]
y=[78,92,65,88,75]
plt.bar(x,y)
plt.title("Student's Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()



```

```python
#ques2: Plot a bar chart for the given data . Change the bar color to green and also add grid lines

months=["Jan","Feb","Mar","Apr","May"]
sales=[250,320,280,400,350]
plt.bar(months,sales,color="green")
#Add title and labels
plt.title("Monthly Sales")           
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)       #add grid lines
plt.show()
```

```python
#q3: Plot the bar chart for the given data . Display the vote value above each bar.

languages=["Python","Java","C++","Javascript"]
votes=[40,25,15,20]
plt.bar(languages,votes)
plt.title("Favorite Programming Languages")
plt.xlabel("Programming languages")
plt.ylabel("Votes")

#display values on top of each bar
for i in range (len(languages)):
    plt.text(i, votes[i],str(votes[i]),ha="center", va="bottom")
plt.show()
```

```python
#q4: Create a bar chart for the following data. Rotate x-axis labels by 45 degree and also add a title
#Data
products=["Laptop","Phone","Tablet","Headphones"]
prices=[55000,30000,25000,50000]
plt.figure(figsize=(6,4))                   #set figure size
plt.bar(products,prices,color="orange")  #create bar chart 
plt.title("Product Prices")
plt.xlabel("Products")
plt.ylabel("Prices")
plt.xticks(rotation=45)               #Rotate x-axis labels
plt.show()

```

```python
#Q6: Create a bar graph for the given data. Change the bar width and also add labels for each bar
#Data
cities=["Delhi","Mumbai","Chennai", "Kolkata"]
population=[19,20,10,15]
plt.bar(cities, population,color="red",width=0.5)
plt.title("Population Comparison")
plt.xlabel("Cities")
plt.ylabel("Population")

#Add labels on top of each bar
for i in range(len(cities)):
    plt.text(i,population[i], str(population[i]),ha="center", va="bottom")
plt.show()
```

```python
#Q7: Create a grouped bar chart for the given data . also add a legend
import matplotlib.pyplot as plt
import numpy as np
subjects=["Math","Science","English"]
boys=[80,75,90]
girls=[85,88,92]
x=np.arange(len(subjects))              #positin of bars
width=0.35

#Create bars
plt.bar(x-width/2,boys,width,label="Boys")
plt.bar(x+width/2,girls,width,label="Girls")

#Labels and title
plt.title("Marks Comparison: Boys vs Girls")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.xticks(x,subjects)             #Subject names on x-axis
plt.legend()                     #show legend
plt.show()
```

```python
#Q9: Plot the temperatures as a bar chart. Highlight the highest temperature bar in red 
days=["Mon","Tues","Wed","Thurs","Fri"]
temp=[30,32,31,33,29]

#Default colors for all bars
colors=["skyblue"]*len(temp)

#highlight the highest temperature in red
highest_index=temp.index(max(temp))
colors[highest_index]="red"

plt.bar(days,temp,color=colors)
plt.title("Temperature analysis")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.show()
```
