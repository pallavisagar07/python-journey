```python
#Q1. Create two 2 × 2 matrices and calculate their matrix multiplication using @.
import numpy as np
a = np.array([[1, 2],[3, 4]])
b = np.array([[5, 6],[7, 8]])
result = a @ b
print(result)
```

```python
#Q2. Calculate the dot product of:

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
result = np.dot(a, b)
print(result)
```

```python
#Q3. Find the transpose of:

a=np.array([[1, 2],[3, 4]])
print(np.transpose(a))
```

```python
#Q4. Find the determinant of:

arr=np.array([[4, 7],[2, 6]])
print(np.linalg.det(arr))
```

```python
#Q5. Find the inverse of:

arr=np.array([[4, 7],[2, 6]])
print(np.linalg.inv(arr))
```

```python
#Q6. Verify that a matrix multiplied by its inverse gives the identity matrix.
import numpy as np
arr = np.array([[4, 7],[2, 6]])
inverse = np.linalg.inv(arr)
print("Inverse:")
print(inverse)

print("Matrix × Inverse:")
print(arr @ inverse)
```

```python
#Q7. Find the eigenvalues of:

import numpy as np
arr = np.array([[2, 0],[0, 3]])
eigenvalues = np.linalg.eigvals(arr)
print(eigenvalues)
```

```python
#Q8. Solve the following system of equations using NumPy:
'''2x + y = 5
x + 3y = 6'''

import numpy as np
A = np.array([[2, 1],[1, 3]])
B = np.array([5, 6])
solution = np.linalg.solve(A, B)
print(solution)
```
