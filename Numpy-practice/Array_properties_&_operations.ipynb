Array Properties

```python
'''Create this array:
arr = np.array([[10, 20, 30],[40, 50, 60]])
Then answer:
Q1. Find its number of dimensions.'''
import numpy as np
arr=np.array([[10,20,30],[40,50,60]])
print(arr)
print()
print(arr.ndim)
```

```python
#2.find its shape
print(arr.shape)
```

```python
#3Find the total number of elements in arr.
print(arr.size)
```

```python
#4 Find the data type of arr.
print(arr.dtype)
```

```python
#Q5. Find the size of each element in bytes.
print(arr.itemsize)
```

```python
#Q6. Create a 3D array and find its:dimensions,shape,size,data type
arr1=np.array([[[7,8,9],[2,5,6]]])
print(arr1)
print(arr1.ndim)              #dimensions of array
print()
print(arr1.shape)              #shape of array
print()
print(arr1.size)            #size of array
print(arr1.dtype)           #type of array
```

```python
'''Use:
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])'''
#Q1. Add a and b.
import numpy as np
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])
varadd=np.add(a,b)
print(varadd)
```

```python
#Q2. Subtract b from a.
sub=np.subtract(a,b)
print(sub)
```

```python
#Q3. Multiply the arrays.
multi=np.multiply(a,b)
print(multi)
```

```python
#Q4. Divide a by b.
div=np.divide(a,b)
print(div)
```

```python
#Q5. Find a².
a = np.array([10, 20, 30, 40, 50])
print(a**2)
```

```python
#Q6. Add 10 to every element of a.
a = np.array([10, 20, 30, 40, 50])
print(a+10)
```

```python
#Q7. Multiply every element of a by 3.
a = np.array([10, 20, 30, 40, 50])
print(a*3)
```

```python
#Q8. Compare a > 25.
a = np.array([10, 20, 30, 40, 50])
print(a > 25)
```

```python
#Q9. Find which elements of a are equal to 30.
a = np.array([10, 20, 30, 40, 50])
print(a[a == 30])
```

```python
#Q10. Calculate the percentage increase between corresponding values of a and b
a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])
percentage = ((a - b) / b) * 100
print(percentage)
```
