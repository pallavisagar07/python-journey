```python
#Use: marks = np.array([65, 78, 90, 56, 88, 72, 95, 45, 81, 69])

#Q1. Find the total marks.
import numpy as np
marks = np.array([65, 78, 90, 56, 88, 72, 95, 45, 81, 69])
print(marks)
print("Total marks:",np.sum(marks))
```

```python
#Q2. Find the average marks.
print("Average marks:",np.mean(marks))
```

```python
#Q3. Find the median.
print("Median is:", np.median(marks))
```

```python
#Q4: Find the minimum marks.
print("Minimum marks:", np.min(marks))
```

```python
#Q5. Find the maximum marks.
print("Maximum marks:", np.max(marks))
```

```python
#Q6. Find the standard deviation.
print("Standard deviation:", np.std(marks))
```

```python
#Q7. Find the variance
print("Variance:", np.var(marks))
```

```python
#Q8. Round the values to 2 decimal places
arr = np.array([10.234, 20.678, 30.456, 40.999])
print(np.round(arr, 2))
```

```python
#Q9: Find the square root of:16, 25, 36, 49, 64
arr=np.array([16, 25, 36, 49, 64])
print(np.sqrt(arr))
```

```python
#Q10 — Find the absolute values of:[-10, 20, -30, 40, -50]

arr = np.array([-10, 20, -30, 40, -50])
print(np.abs(arr))
```
