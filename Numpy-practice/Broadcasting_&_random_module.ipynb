```python
#Q1. Add 10 to every element:
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(arr+10)
```

```python
#Q2. Multiply every element by 5.
arr = np.array([1, 2, 3, 4, 5])
print(arr*5)
```

```python
#Q3. Add:
a = np.array([10, 20, 30])
b = 5
print(a+b)
```

```python
#Q4. Add these arrays using broadcasting:
a = np.array([[1, 2, 3],[4, 5, 6]])
b = np.array([10, 20, 30])
print(a+b)
```

```python
#Q5. Subtract [5, 10, 15] from every row of:
arr = np.array([[20, 30, 40],[50, 60, 70]])
print(arr - np.array([5, 10, 15]))
```

```python
#Q21. Increase every student's marks by 5:
marks = np.array([[70, 80, 90],[60, 75, 85],[88, 92, 95]])
print(marks+5)
```

```python
#Q1. Generate 5 random numbers between 0 and 1.
import numpy as np
var = np.random.random(5)
print(var)
```

```python
#Q2. Generate 10 random integers between 1 and 100.
var1=np.random.randint(1,101,10)
print(var1)
```

```python
#Q3. Generate a 3 × 4 array of random integers between 10 and 50.
arr = np.random.randint(10, 51, (3, 4))
print(arr)
```

```python
#Q4. Generate 5 random numbers from a normal distribution.
var = np.random.normal(size=5)
print(var)
```

```python
#Q5.Set a random seed and generate 5 random integers. 
#Run the code again and verify that you get the same values.
import numpy as np
np.random.seed(42)
arr = np.random.randint(1, 101, 5)
print(arr)
```

```python
#Q6. Create an array:[10, 20, 30, 40, 50].Randomly shuffle it.
arr=np.array([10,20,30,40,50])
np.random.shuffle(arr)
print(arr)
```

```python
#Q7. Randomly select 3 values from:[10, 20, 30, 40, 50, 60, 70]
arr2=np.array([10, 20, 30, 40, 50, 60, 70])
result = np.random.choice(arr2, 3)
print(result)
```
