```python
#Q10: count number of vowels and consonants in string
text=input("Enter string")
#initialize counters
vowels=0
consonants=0
for ch in text.lower():
        if ch in"aeiou":
            vowels+=1
        else:
            consonants+=1

print("Vowels:", vowels)
print("Consonants:", consonants)
```
